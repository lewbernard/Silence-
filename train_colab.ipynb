# Train the ASL motion model on Google Colab

This notebook runs the repository's `train.py` using a Colab GPU when one is available. The model checkpoint is downloaded to your computer when training finishes.

Before starting, make sure `asl_motion_sequence_dataset.csv` contains examples for all 26 letters, recorded across at least two separate collection sessions.

## 1. Select a GPU runtime

In Colab, open **Runtime → Change runtime type** and select a GPU accelerator, then save. GPU availability depends on your Colab account and current resource limits.

```python
import torch

if torch.cuda.is_available():
    print(f"GPU ready: {torch.cuda.get_device_name(0)}")
else:
    print("No GPU detected. Training will still run, but it will use the CPU and take longer.")
```

## 2. Upload the training files

Run the cell below and upload each requested file separately. Do not upload `requirements.txt`: Colab already provides PyTorch, and the repository's pinned local packages are not needed for this training script.

```python
from google.colab import files
from pathlib import Path

required_files = (
    "train.py",
    "sequence_dataset.py",
    "sequence_model.py",
    "asl_motion_sequence_dataset.csv",
)
for filename in required_files:
    if Path(filename).is_file():
        print(f"{filename} is already uploaded; skipping.")
        continue
    print(f"Upload {filename} when the file picker opens.")
    uploaded = files.upload()
    if filename not in uploaded:
        raise FileNotFoundError(
            f"Expected {filename}; please rerun this cell and select that file."
        )
print("All required files are ready.")
```

## 3. Train

Training evaluates both the GRU and temporal CNN, then saves the best checkpoint as `asl_motion_model.pth`. It may take a while; keep the Colab tab/runtime active until it finishes.

```python
!python train.py
```

## 4. Download the trained model

Keep this checkpoint with your project, replacing the existing `asl_motion_model.pth` if you want to use the newly trained model. Colab's temporary files disappear when the runtime is deleted.

```python
from pathlib import Path
from google.colab import files

checkpoint = Path("asl_motion_model.pth")
if not checkpoint.is_file():
    raise FileNotFoundError("Training did not produce asl_motion_model.pth")
files.download(str(checkpoint))
```
